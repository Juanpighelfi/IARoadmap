```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Grafica las 3 funciones de activación (sigmoid, relu, tanh) en un solo subplot
#    con leyenda, grilla, y colores distintos. Rango x: [-6, 6]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
# axes es un array de 3 ejes

x = np.linspace(-5, 5, 100)

# Subplot 1: Sigmoid
axes[0].plot(x, 1 / (1 + np.exp(-x)), color='blue', linewidth=2)
axes[0].set_title('Sigmoid')
axes[0].axhline(y=0.5, color='gray', linestyle='--', alpha=0.5)

# Subplot 2: ReLU
axes[1].plot(x, np.maximum(0, x), color='green', linewidth=2)
axes[1].set_title('ReLU')

# Subplot 3: Tanh
axes[2].plot(x, np.tanh(x), color='red', linewidth=2)
axes[2].set_title('Tanh')

for ax in axes:
    ax.grid(True, alpha=0.3)
    ax.set_xlabel('x')

plt.suptitle('Funciones de Activación', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python

# 2. Genera un "dataset" de clasificación 2D:
#    - Clase 0: 200 puntos centrados en (2, 2) con std=1
#    - Clase 1: 200 puntos centrados en (-2, -2) con std=1
#    Grafica con scatter, colores distintos por clase


# Clase 0: 200 puntos centrados en (2, 2) con std=1
# Usamos tu línea correcta para x0 y y0
x0 = np.random.normal(loc=2, scale=1, size=200) # Coordenadas x de la Clase 0
y0 = np.random.normal(loc=2, scale=1, size=200) # Coordenadas y de la Clase 0

# Clase 1: 200 puntos centrados en (-2, -2) con std=1
# Ahora cambiamos el centro (media) a -2
x1 = np.random.normal(loc=-2, scale=1, size=200) # Coordenadas x de la Clase 1
y1 = np.random.normal(loc=-2, scale=1, size=200) # Coordenadas y de la Clase 1

# --- PASO 3: Graficar con Matplotlib ---

# Creamos una figura
plt.figure(figsize=(8, 6))

# Graficamos la Clase 0 con scatter. Parámetros: x, y, color, etiqueta
plt.scatter(x0, y0, color='blue', label='Clase 0 (centro (2,2))', alpha=0.6)

# Graficamos la Clase 1. Usamos otro color.
plt.scatter(x1, y1, color='red', label='Clase 1 (centro (-2,-2))', alpha=0.6)

# Detalles estéticos para que se entienda el gráfico
plt.title('Dataset de Clasificación 2D Sintético')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.legend() # Muestra la leyenda con los label
plt.grid(True, linestyle='--', alpha=0.7) # Cuadrícula de fondo

# Aseguramos que los ejes se vean centrados para apreciar la simetría
plt.axhline(0, color='black',linewidth=1)
plt.axvline(0, color='black',linewidth=1)

# Mostramos el gráfico
plt.show()
```

```python
# 3. "Simula" el entrenamiento de un modelo:
#    - Genera 50 valores de loss que van de 2.5 a 0.05 con algo de ruido
#      (pista: usa np.linspace + np.random.randn * factor_decreciente)
#    - Genera 50 valores de accuracy que van de 0.5 a 0.97
#    - Crea una figura con 2 subplots: loss y accuracy vs epoch
#    - Marca el epoch del mejor accuracy con una línea vertical roja
values = np.linspace(2.5, 0.05, 50)
noise = np.random.normal(loc=0, scale=0.001, size=50)
data = values + noise
accuracy = np.linspace(0.5, 0.97, 50)

fig, axes = plt.subplots(1,2)
axes[0].plot(np.arange(50), data)
axes[1].plot(np.arange(50), accuracy)

mejor_epoch = np.argmax(accuracy)

# 2. Dibujamos la línea en el segundo subplot (axes[1])
axes[1].axvline(x=mejor_epoch, color='red', linestyle='--', label='Mejor Accuracy')
plt.show()


```
