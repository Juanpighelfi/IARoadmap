

## ⏱️ Cronograma del Sprint (60 min)

### 1. La Anatomía de una Figura (10 min)
Matplotlib no es solo "dibujar líneas", es una jerarquía de objetos.
* **Concepto clave:** La diferencia entre `Figure` (el lienzo total) y `Axes` (el gráfico individual donde viven los datos).
* **Interfaz:** Usaremos siempre el enfoque orientado a objetos (`fig, ax = plt.subplots()`), que es más robusto que el estilo simple de MATLAB.



> **🧩 Desafío Socrático 1:** Si creas una figura con `plt.subplots(2, 2)`, ¿qué tipo de objeto devuelve la variable `ax` y cómo accederías al gráfico de la esquina inferior derecha?

### 2. Los "Big Three" y Broadcasting (15 min)
Graficar datos de NumPy. 
* **Line Plot:** `ax.plot(x, y)` para series temporales (ej. velocidad en una vuelta).
* **Scatter Plot:** `ax.scatter(x, y)` para correlaciones.
* **Bar Plot:** `ax.bar(categories, values)` para comparar features del ERP.

**💡 Tip Pro:** Matplotlib entiende arrays de NumPy nativamente. Si pasas una matriz de `(100, 5)` a `ax.plot()`, graficará 5 líneas automáticamente.

### 3. Estética y Legibilidad (15 min)
Un gráfico sin etiquetas es solo ruido.
* `ax.set_title()`, `ax.set_xlabel()`, `ax.set_ylabel()`.
* Uso de `label` en el plot y `ax.legend()` para identificar series.
* Personalización: colores (`c='red'`), estilos de línea (`ls='--'`) y transparencia (`alpha=0.5`).

> **🧩 Desafío Socrático 2:** ¿Por qué es mejor práctica usar `ax.set_title()` en lugar de `plt.title()` cuando trabajas con múltiples subplots?

### 4. Layouts Complejos: Subplots (10 min)
Comparar diferentes métricas en una sola imagen.
* `plt.tight_layout()`: El comando "mágico" para que las etiquetas no se encimen.
* Compartir ejes: `plt.subplots(..., sharex=True)` (útil para comparar telemetría en el mismo eje de tiempo).

## 🏎️ Práctica Final: El "Telemetría Dash" (10 min)

Imagina que tienes datos de una frenada en iRacing. Escribe un script que haga lo siguiente:

1.  Crea un array `t` (tiempo) de 0 a 5 segundos con 100 puntos.
2.  Crea un array `v` (velocidad) que empiece en 200 y baje linealmente a 50.
3.  Crea un array `p` (presión de freno) que sea 0 al inicio, suba a 100 y baje.
4.  **El reto:** Crea una figura con **2 subplots verticales** que compartan el eje X.
    * Arriba: La velocidad en una línea roja.
    * Abajo: La presión de freno como un área sombreada (pista: busca `ax.fill_between`).
